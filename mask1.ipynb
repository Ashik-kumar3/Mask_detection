```python
x = []
y = []
data_path = r'C:\Users\ASHIK KUMAR\Desktop\AI\AI\DL\CNN\data'
categories = ['with_mask', 'without_mask']
```

```python
for i in categories:
    path = os.path.join(data_path,i)
    print(i)
    for img in os.listdir(path):
        img_array = imread(os.path.join(path,img))
        img_array = resize(img_array,(150,150,1))
        x.append(img_array)
        y.append(categories.index(i))
```

```python
x = np.array(x)
y = np.array(y)
y
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest = train_test_split(x,y,test_size=.2,random_state=0)
xtrain
```

```python
model = Sequential()
```

```python
model.add(Conv2D(3,(3,3),activation='relu',input_shape=(150,150,1)))
model.add(MaxPool2D(pool_size=(2,2)))
model.add(Flatten())
model.add(Dense(100,activation='relu'))
model.add(Dense(100,activation='relu'))
model.add(Dense(2,activation='softmax'))
```

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model.fit(xtrain,ytrain,epochs=10,batch_size=200)
```

```python
model.evaluate(xtest,ytest,batch_size=100)
```

```python
img2 = imread('masked_man.jpg')
img2 = resize(img2,(150,150,1))
img2 = np.array([img2])
```

```python
pred = model.predict(img2)
pred = np.argmax(pred)
pred = categories[pred]
pred
```
